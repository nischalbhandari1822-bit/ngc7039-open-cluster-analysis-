```python
# Unit conversion
x = 1.0 * u.parsec
x.to(u.km) # (syntex: object.to(u.Unit))
```

```python
from astropy.units import imperial

cms = u.cm / u.s
# ...and then use some imperial units
mph = imperial.mile / u.hour

# And do some conversions
q = 42.0 * cms
q.to(mph)
```

```python
# Most important is Equivalencies 
(1000 * u.nm).to(u.Hz)
# regularly if we try to convert units of two different quantities it doesnot give us the value.
```

```python
# Performance enhancement tips
import numpy as np
from astropy import units as u

rng = np.random.default_rng()
array = rng.random(10000000)

%timeit array * u.m / u.s / u.kg / u.sr
```

```python
# Faster alternative:
UNIT_MSKGSR = u.m / u.s / u.kg / u.sr

%timeit array * UNIT_MSKGSR
```
