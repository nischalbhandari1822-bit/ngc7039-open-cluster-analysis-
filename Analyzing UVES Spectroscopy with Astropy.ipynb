UVES spectroscopy refers to analyzing high-resolution data from the UV-Visual Echelle Spectrograph(UVES). It splits incoming celestial light into two paths; a blue arm(300 to 500 nm) and a red arm(420 to 1100nm) capturing everything from ultravoilet to near-infrared light. It yields an incredibly high resolving power (up to 110,000). This allows astronomers to see minuscale "fingerprints" of chemical elements in stars and distant galaxies.

Backgound: We are up to analyze data from  MN lup, a T Tauri star in the Taurus-Auriga star forming region located at a distance of about 140pc. 

Here we are analyzing the UVES data to search for rotationally modulated features in the emission line profiles, which could be due to an accretion spot on the stellar surface. 
Rotationally modulated features in stellar emission line profiles are periodic changes in spectral lines such as H alpha, Ca II, UV wind lines, caused by the rotation of  non uniform surface structures, magnetic fields, or outflowing wind features across the visible star.
