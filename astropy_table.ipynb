```python
from astropy.table import QTable
import astropy.units as u 
import numpy as np 

a= np.array([1,4,5], dtype=np.int32)# here even if the data type is not included, default will be the same int32
b=[2.0,5.0,8.5]
c=['x','y', 'z']
d= [10,20,30]* u.m/u.s # Quantity array(Anything associated with units)

t=QTable([a,b,c,d],
          names=('a','b','c','d'),
          meta={'name':'first table'})
t
#t.show_in_browser()
#t.show_in_browser(jsviewer=True)
```

```python
from astropy.table import MaskedColumn
a_masked = MaskedColumn(a, mask=[True, True, False])
t = QTable([a_masked, b, c], names=('a', 'b', 'c'),
           dtype=('i4', 'f8', 'U1'))
t
```

```python
# For the Missing Values

from astropy.table import MaskedColumn
a_masked = MaskedColumn(a, mask=[True, True, False])
t = QTable([a_masked, b, c], names=('a','b','c'),
           dtype=('i4','f8','U1'),
           )

t
```

```python
from astropy.time import Time
from astropy.coordinates import SkyCoord

tm = Time(['2000:002', '2002:345'])
sc = SkyCoord([10, 20], [-45, +40], unit='deg')
t = QTable([tm, sc], names=['time', 'skycoord'])
t
```

```python
dt=t['time']- Time('1990-07-23 04:30:59.984')
print(dt)
```

```python
# dt_cxo  means date(time interval) since the launch of Chandra X ray Observatory aboard STS-93
# units are set at days. (3449 means nearly 10 years)
t['dt_cxo']= dt.to(u.d)
t['dt_cxo'].info.format= '.3f'
print(t)
```

```python
# Constructing tables row by row is slow, so performance tricks:
 
from astropy.table import Table
t= Table(names=['a','b'])
for i in range(100): # range = no. of rows
    t.add_row((1,2))
#t
```

```python
from astropy.table import Table
import numpy as np

x = np.arange(10000, dtype=float)
tm = Table([x], masked=True)
tm.write('tm.ecsv', overwrite=True)
x
```

```python
# read() takes a lot of time. 
# in some cases it is possible to only read a subset of the table by choosing the option(memmap=True)

# this method is most important while reading fits binary table
#it is possible to read only a subset of rows, but reading a full column loads the whole table data into memory:

import numpy as np
from astropy.table import Table

tbl= Table({'a':np.arange(1e7),
            'b': np.arange(1e7, dtype='float'),
            'c': np.arange(1e7, dtype='float')})
#tbl
# just for making a table 
```

```python
tbl.write('test.fits', overwrite=True)
table = Table.read('test.fits', memmap=True)
table2=tbl[:100]
print(table2)
# basically (memmap=True) says I need up to 100 rows just read up to 100 and give me the table

```
